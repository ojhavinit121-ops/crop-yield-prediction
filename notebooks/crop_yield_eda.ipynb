# Crop Yield Prediction — Exploratory Data Analysis
Initial exploration of the supplied modeling dataset.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('../data/yield_features.csv')
df.head()
```

## Dataset overview

```python
df.shape
df.info()
```

## Missing values and duplicates

```python
display(df.isnull().sum().sort_values(ascending=False))
print('Duplicate rows:', df.duplicated().sum())
```

## Yield distribution

```python
target = 'hg/ha_yield'
display(df[target].describe())
df[target].plot(kind='hist', bins=40, title='Distribution of crop yield')
plt.xlabel('Yield (hg/ha)')
plt.show()
```

## Average yield by crop

```python
df.groupby('Item')[target].mean().sort_values(ascending=False).plot(kind='bar', figsize=(10, 5))
plt.ylabel('Average yield (hg/ha)')
plt.tight_layout()
plt.show()
```
